```python
train ={
    "Train_Number": 12951,
    "Train_Name": "Mumbai Rajdhani Express",
    "Source": "Mumbai",
    "Destination":"Delhi",   "Available_seats":120
}
print(train)
type(train)
print("\nTrain Name:", train["Train_Name"])
print("Train Number:", train["Train_Number"])
print("Source Station:", train["Source"])
print("Distination Station:"), train["Destination"]
print("Available Seats:", train["Available_seats"])
train["Available_seats"] = 125
print(train)

print("\n\n\nTrain Name:", train["Train_Name"])
print("Train Number:", train["Train_Number"])
print("Source Station:", train["Source"])
print("Distination Station:"), train["Destination"]
print("Updated Seats:", train["Available_seats"])


train.update({"Platform": 4})
print("\nTrain Name:", train["Train_Name"])
print("Train Number:", train["Train_Number"])
print("Source Station:", train["Source"])
print("Distination Station:"), train["Destination"]
print("Available Seats:", train["Available_seats"])
print("Platform Number:", train["Platform"])

train.pop("Platform")
print("\n\n\nTrain Name:", train["Train_Name"])
print("Train Number:", train["Train_Number"])
print("Source Station:", train["Source"])
print("Distination Station:"), train["Destination"]
print("Updated Seats:", train["Available_seats"])


print("\n\nUpdated Train Information:")
print("Train Name:", train["Train_Name"])
print("Train Number:", train["Train_Number"])
print("Source Station:", train["Source"])
print("Distination Station:"), train["Destination"]
print("Updated Seats:", train["Available_seats"])
```
