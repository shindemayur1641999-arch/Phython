```python
num = int(input("Enter a number: "))

fact = 1

for i in range(1, num + 1):
    fact = fact * i

print("Factorial =", fact)
```

```python
def fact(n):
  if n == 0 or n == 1:
     return 1
  return n * fact(n-1)

num = 3
print("fact:", fact(num))
```
