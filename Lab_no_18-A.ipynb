```python
import re

class RailwayReservationSystem:

    def __init__(self):
        self.train_name = "Rajdhani Express"
        self.available_seats = 5
        self.passengers = []

    def book_ticket(self, name, pnr, seat, fare):
        if self.available_seats > 0:
            passenger = {
                "name": name,
                "pnr": pnr,
                "seat": seat,
                "fare": fare
            }
            self.passengers.append(passenger)
            self.available_seats -= 1
            return f"Ticket booked for {name} on seat {seat}."
        else:
            return "No seats available. Booking failed."

    def validate_ticket(self, pnr, seat):

        pnr_pattern = r'^\d{8}$'
        seat_pattern = r'^[A-Z][0-9]{1,2}$'

        pnr_valid = re.fullmatch(pnr_pattern, pnr)
        seat_valid = re.fullmatch(seat_pattern, seat)

        return bool(pnr_valid and seat_valid)

    def allocate_seat(self):
        # This method is designed to be called separately to simulate seat allocation management.
        # If book_ticket already handles seat assignment, this could be for a different purpose
        # like re-assigning seats or allocating from a general pool without specific booking info.
        if self.available_seats > 0:
            # In a real system, you might allocate a specific seat number here.
            # For this example, we just decrement the count and return a success message.
            self.available_seats -= 1
            return "Seat successfully allocated."
        else:
            return "No seats left to allocate."
```

```python
system = RailwayReservationSystem()
```

```python
system.book_ticket(
   "Rahul Sharma",
   "43821756",
   "A1",
   1200
)

system.book_ticket(
   "Priya Singh",
   "67934521",
   "A2",
   1200
)

print(system.passengers)
```

```python
for passenger in system.passengers:

   result = system.validate_ticket(
       passenger["pnr"],
       passenger["seat"])

   print(
       passenger["name"],
       "-",
       result)
```

```python
for passenger in system.passengers:

   print(
       passenger["name"],
       "-",
       system.allocate_seat())
```

```python
print(
   "Available Seats :",
   system.available_seats)
```

```python
for passenger in system.passengers:

    print("=" * 40)

    print("Passenger :", passenger["name"])
    print("PNR       :", passenger["pnr"])
    print("Seat      :", passenger["seat"])
    print("Fare      :", passenger["fare"])

    print("=" * 40)

print("\nBooking Summary")
print("-" * 30)

print(
    "Total Passengers :",
    len(system.passengers))

print(
    "Available Seats :",
    system.available_seats)

if system.available_seats > 0:
    print("Reservation Status : Open")
else:
    print("Reservation Status : Full")
```
